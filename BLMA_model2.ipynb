```python
import pandas as pd

df_income['date_census'] = pd.to_datetime(
    df_income['date_census'],
    format='%Y-%m-%d',
    errors='coerce'
)
df_income['date_census'].dtype
```

```python
df_income.head(15)
```

```python
df_income = df_income.sort_values(['zcta', 'date_census']).copy()
```

```python
df_income['year'] = df_income['date_census'].dt.year
```

```python
df_income['target_mlp_1y'] = (
    df_income
    .groupby('zcta')['median_listing_price']
    .shift(-1)
)
```

```python
df_income['mlp_lag1'] = (
    df_income
    .groupby('zcta')['median_listing_price']
    .shift(1)
)

df_income['mlp_lag2'] = (
    df_income
    .groupby('zcta')['median_listing_price']
    .shift(2)
)
```

```python
df_model = df_income.dropna(
    subset=[
        'median_listing_price',
        'target_mlp_1y',
        'mlp_lag1'
    ]
).copy()
```

```python
df_model.head(15)
```

```python
train_df = df_model[df_model['year'] <= 2021].copy()
val_df   = df_model[df_model['year'] == 2022].copy()

pred_2024_df = df_income[df_income['year'] == 2023].copy()
```

```python
features = [
    'zcta',
    'year',
    'median_listing_price',
    'mlp_lag1',
    'household_income',
    'household_income_prior',
    'household_income_delta'
]

target = 'target_mlp_1y'
```

```python
from catboost import CatBoostRegressor
```

```python
cat_features = ['zcta']

model = CatBoostRegressor(
    iterations=800,
    learning_rate=0.05,
    depth=8,
    loss_function='RMSE',
    random_seed=42,
    verbose=100
)

model.fit(
    train_df[features],
    train_df[target],
    cat_features=cat_features,
    eval_set=(val_df[features], val_df[target]),
    use_best_model=True
)
```

```python
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error
import numpy as np

val_preds = model.predict(val_df[features])

rmse = np.sqrt(mean_squared_error(val_df[target], val_preds))
mape = mean_absolute_percentage_error(val_df[target], val_preds)

print("Validation RMSE:", rmse)
print("Validation MAPE:", mape)
```

```python
pred_2024_df['all_homes_projections'] = model.predict(
    pred_2024_df[features]
)
```
